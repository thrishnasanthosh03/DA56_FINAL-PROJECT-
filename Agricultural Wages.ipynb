 # Dataset Overview & Introduction
This dataset contains 500 records and 14 columns detailing rural agricultural wage metrics across different regions, states, districts, and collection centers. It captures temporal, spatial, and demographic data—including labor categories, specific agricultural work types, gender breakdowns, and both monthly and annual average wages.   


## 2. Dataset Schema & Structure
The dataset (`Agricultural_Wages_Sample_Data.csv`) contains **500 records** and **14 columns** structured as follows:

| Column Name | Description |
| :--- | :--- |
| **Date (month)** | Recording month and year |
| **State Name / Code** | State or Union Territory identification |
| **District Name / Code** | District-level geographic mapping |
| **Center Name / Code** | Data collection center and coordinates (Latitude/Longitude) |
| **Category & Type of Labour** | Classification of agricultural work tasks |
| **Gender of Labour** | Workforce demographic categorization |
| **Monthly & Annual Wages** | Financial wage metrics |

```python
df['date_(month)'] = pd.to_datetime(df['date_(month)'])
```

Converts the data type of the date_(month) column from a plain string/text object (object) into a standard Pandas datetime format (datetime64).

Exploratory Data Analysis (EDA)

```python
df['year'] = df['date_(month)'].dt.year
df['month'] = df['date_(month)'].dt.month
```

created year,month seperate columns

```python
df['latitude_(latitude)'] = df['latitude_(latitude)'].replace(0.0, None)
df['longitude_(longitude)'] = df['longitude_(longitude)'].replace(0.0, None)
df['latitude_(latitude)'] = df['latitude_(latitude)'].ffill()
df['longitude_(longitude)'] = df['longitude_(longitude)'].ffill()
```

Replaces invalid 0.0 coordinates in the latitude and longitude columns with missing values (None), and then fills those gaps using the previous valid value (ffill).

```python
df['average_monthly_wage_(monthly_average_wage)'] = df['average_monthly_wage_(monthly_average_wage)'].replace(0.0, None)
df['annual_average_wage_(annual_average_wage)'] = df['annual_average_wage_(annual_average_wage)'].replace(0.0, None)

monthly_median= df['average_monthly_wage_(monthly_average_wage)'].median()
df['average_monthly_wage_(monthly_average_wage)'] = df['average_monthly_wage_(monthly_average_wage)'].fillna(monthly_median)

df['annual_average_wage_(annual_average_wage)']=df['annual_average_wage_(annual_average_wage)'].fillna(df['average_monthly_wage_(monthly_average_wage)']*12)
```

```python
df.head(10)
```

Replaces erroneous 0.0 values in both monthly and annual wage columns with NaN, fills missing monthly wages using the column's median, and correctly computes the corresponding annual wages (monthly wage * 12) to ensure full data integrity.

```python
bins = [0, 200, 500, df['average_monthly_wage_(monthly_average_wage)'].max()]
labels = ['Low Wage', 'Medium Wage', 'High Wage']
df['wage_category'] = pd.cut(df['average_monthly_wage_(monthly_average_wage)'], bins=bins, labels=labels)
```

```python
df[['average_monthly_wage_(monthly_average_wage)', 'wage_category']].head(10)
```

```python
print(df['wage_category'].value_counts())
```

Segments continuous monthly wages into discrete bins (0–200, 200–500, and up to the maximum value) and assigns readable labels (Low Wage, Medium Wage, High Wage).
