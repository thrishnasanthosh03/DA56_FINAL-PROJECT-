## Introduction to Agricultural Wages Analysis

**Agricultural wages** show the daily and monthly earnings of workers engaged in various farming activities across different states and districts in India. The study of agricultural wages is critical to understanding rural livelihoods, income disparities, labor mobility, and the economic well-being of agricultural households.

The purpose of this analysis is to examine trends in agricultural wages, compare wages across states, gender categories, and labor types, and present key insights through clear and meaningful data visualization. This project uses Python libraries such as **Pandas, NumPy, Plotly, Matplotlib, and Seaborn** to analyze an agricultural wages dataset.

## Data Source

The dataset used in this project covers detailed records of agricultural wage rates collected across various centers in India.
**Note:-** The dataset structure and field parameters remain original and preserved for comprehensive analysis.

## Attribute Details

| Attribute Name | Description |
| :--- | :--- |
| **Date (month)** | The date and month recording when the wage data was collected. |
| **State Name (state_name)** | Represents the name of the state in India where the wage data was recorded. |
| **State Code (state_code)** | Unique numerical identification code assigned to the state. |
| **District Name (district_name)** | Represents the name of the district within the state. |
| **District Code (district_code)** | Unique numerical identification code assigned to the district. |
| **Center Name (center_name)** | The specific data collection center or village/town location name. |
| **Center Code (center_code)** | Unique numerical identification code assigned to the center location. |
| **Latitude (latitude)** | Geographic coordinate specifying the north-south position of the center. |
| **Longitude (longitude)** | Geographic coordinate specifying the east-west position of the center. |
| **Category of labour (labour_category)** | Specifies the broad classification of labor (e.g., Field Labour, Skilled Labour, Other Agricultural Labour). |
| **Labour work engaged in (labour_type)** | Specifies the exact agricultural operation performing (e.g., Sower, Weeder, Harvester, Blacksmith, Carpenter). |
| **Gender of Labour (gender)** | Indicates the gender category of the worker (e.g., Men, Women). |
| **Average monthly wage (monthly_average_wage)** | Average wage rate recorded for the worker on a monthly basis. |
| **Annual average wage (annual_average_wage)** | Calculated average wage rate of the worker across the year. |

**Note:-** *Zero values in average monthly wage indicate missing or non-recorded entries for that specific period or category.*
**Info:-** *This dataset consists of 500 records and 14 attributes, providing a structured collection of geographical, labor category, and wage level information that supports effective agricultural wage analysis.*

### Creating Derived Columns

```python
df['year'] = df['date_(month)'].dt.year
df['month'] = df['date_(month)'].dt.month
```

***info***:- created year,month seperate columns

### Replaces erroneous 0.0 values

```python
df['latitude_(latitude)'] = df['latitude_(latitude)'].replace(0.0, None)
df['longitude_(longitude)'] = df['longitude_(longitude)'].replace(0.0, None)
df['latitude_(latitude)'] = df['latitude_(latitude)'].ffill()
df['longitude_(longitude)'] = df['longitude_(longitude)'].ffill()
```

Replaces invalid 0.0 coordinates in the latitude and longitude columns with missing values (None), and then fills those gaps using the previous valid value (ffill).

```python
df['average_monthly_wage_(monthly_average_wage)'] = df['average_monthly_wage_(monthly_average_wage)'].replace(0.0, None)
df['annual_average_wage_(annual_average_wage)'] = df['annual_average_wage_(annual_average_wage)'].replace(0.0, None)

monthly_median= df['average_monthly_wage_(monthly_average_wage)'].median()
df['average_monthly_wage_(monthly_average_wage)'] = df['average_monthly_wage_(monthly_average_wage)'].fillna(monthly_median)

df['annual_average_wage_(annual_average_wage)']=df['annual_average_wage_(annual_average_wage)'].fillna(df['average_monthly_wage_(monthly_average_wage)']*12)
```

```python
df.head(10)
```

***info***- Replaces erroneous 0.0 values in both monthly and annual wage columns with NaN, fills missing monthly wages using the column's median, and correctly computes the corresponding annual wages (monthly wage * 12) to ensure full data integrity.

### Importing Required Libraries
Neccessary Python libraries such as Pandas,Plotly,Matplotli,and Seaborn are imported for numerical operations and data visualizations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

### (i). State-wise Average Monthly Wage

• ***Visualization Type:-*** Bar Chart (Seaborn / Matplotlib) 

• ***Purpose:-*** To compare the average monthly agricultural wage across different states. 

• ***Insight:-*** This visualization clearly highlights regional wage disparities among various states. For instance, states like Kerala exhibit significantly higher average monthly wages compared to others.

```python

# Calculate state-wise average monthly wage
state_wage = df.groupby('state_name_(state_name)')[
    'average_monthly_wage_(monthly_average_wage)'
].mean().reset_index()

state_wage_sorted = state_wage.sort_values(
    by='average_monthly_wage_(monthly_average_wage)', 
    ascending=False
)

# Plot the bar chart
plt.figure(figsize=(12, 6))
sns.barplot(
    data=state_wage_sorted, 
    x='state_name_(state_name)', 
    y='average_monthly_wage_(monthly_average_wage)', 
    palette='viridis'
)
plt.xticks(rotation=90)
plt.title('State-wise Average Monthly Wage')
plt.xlabel('State Name')
plt.ylabel('Average Monthly Wage')
plt.tight_layout()
plt.show()
```

***Info:-*** This bar chart illustrates the average monthly wages across various Indian states within the agricultural sector. The data reveals substantial regional inequality, with states like Kerala and Tripura offering significantly higher monthly earnings, whereas states such as Punjab and Chhattisgarh show much lower average wage rates. This variation highlights key regional economic differences and labor market dynamics across different parts of the country.

### (ii). Year-over-Year Wage Trend (Line Chart)

• ***Visualization Type:-*** Line Chart (Seaborn / Matplotlib)

• ***Purpose:-*** To analyze temporal wage progression and determine whether agricultural wages have increased over the years from 2010 to 2020.   

• ***Insight:-*** The trend clearly shows a steady upward trajectory in average monthly wages over time, reflecting overall growth in rural wage rates.  

```python
 #Extract year from date and calculate year-wise average monthly wage

df['year'] = pd.to_datetime(df['date_(month)']).dt.year
year_trend = df.groupby('year')[
    'average_monthly_wage_(monthly_average_wage)'
].mean().reset_index()

# Plot the line chart
plt.figure(figsize=(10, 5))
sns.lineplot(
    data=year_trend, 
    x='year', 
    y='average_monthly_wage_(monthly_average_wage)', 
    marker='o', 
    color='b', 
    linewidth=2.5
)
plt.title('Year-over-Year Average Monthly Wage Trend')
plt.xlabel('Year')
plt.ylabel('Average Monthly Wage')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

***Info:-*** This line chart tracks the progression of average agricultural monthly wages across multiple years. It demonstrates how wage rates have evolved over time, showing a general pattern of annual growth and economic shifts within the rural labor market.

### (iii). Gender-based Wage Comparison across Labour Categories

• ***Visualization Type:-*** Grouped Bar Chart (Seaborn / Matplotlib)

• ***Purpose:-*** To compare the average monthly wage between male and female workers across different labor categories (e.g., Field Labour, Skilled Labour, Herdsmen).

• ***Insight:-*** This chart evaluates the gender wage gap within the agricultural sector, showing noticeable disparities in earnings between men and women across various operational tasks.

```python
# Calculate average monthly wage grouped by labour category and gender
gender_category_wage = df.groupby([
    'category_of_labour_(labour_category)', 
    'gender_of_labour_(gender)'
])['average_monthly_wage_(monthly_average_wage)'].mean().reset_index()

# Plot the grouped bar chart
plt.figure(figsize=(12, 6))
sns.barplot(
    data=gender_category_wage, 
    x='category_of_labour_(labour_category)', 
    y='average_monthly_wage_(monthly_average_wage)', 
    hue='gender_of_labour_(gender)', 
    palette='Set2'
)
plt.title('Gender-based Wage Comparison across Labour Categories')
plt.xlabel('Labour Category')
plt.ylabel('Average Monthly Wage')
plt.legend(title='Gender')
plt.tight_layout()
plt.show()
```

***info***:- This grouped bar chart highlights gender-based wage variations across distinct categories of agricultural labor. The comparison indicates persistent disparities, where male laborers generally earn higher average monthly wages than female laborers performing similar categories of work.

### (iv). Distribution of Monthly Wages

• ***Visualization Type:-*** Histogram with KDE Plot (Seaborn)

• ***Purpose:-*** To understand the distribution spread of monthly wages and identify which wage amounts are most frequent across the dataset.

• ***Insight:-*** The distribution reveals the concentration of workers across various wage brackets, indicating whether the majority of agricultural laborers fall into lower or higher monthly earning ranges

```python
# Plot the distribution of average monthly wages with KDE curve
plt.figure(figsize=(10, 5))
sns.histplot(
    df['average_monthly_wage_(monthly_average_wage)'], 
    kde=True, 
    color='purple', 
    bins=30
)
plt.title('Distribution of Average Monthly Wages')
plt.xlabel('Average Monthly Wage')
plt.ylabel('Frequency / Count')
plt.tight_layout()
plt.show()
```

***info***:- This histogram combined with a Kernel Density Estimation (KDE) plot illustrates the frequency distribution of average monthly wages. It helps visualize the spread and concentration of earnings, showing how wage values are distributed throughout the agricultural workforce.

### (v). Wage Categories Breakdown

• ***Visualization Type:-*** Pie Chart / Donut Chart (Matplotlib)

• ***Purpose:-*** To analyze the percentage distribution of agricultural workers across different wage categories (Low Wage, Medium Wage, High Wage).

• ***Insight:-*** This chart clearly displays the proportional breakdown of the workforce, highlighting which earning category comprises the majority of agricultural laborers.

```python
# Calculate value counts for each wage category
wage_counts = df['wage_category'].value_counts()

# Plot pie chart
plt.figure(figsize=(6, 7))
plt.pie(
    wage_counts, 
    labels=wage_counts.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=['#ff9999','#66b3ff','#99ff99']
)
plt.title('Wage Categories Breakdown')
plt.tight_layout()
plt.show()
```

***info***:- This pie chart illustrates the percentage breakdown of agricultural workers categorized by wage levels (Low Wage, Medium Wage, and High Wage). It provides a clear visual summary of the overall wage distribution, showing the proportion of workers falling within each specific earning tier.

### (vi). Gender Distribution Across Labour Categories

• ***Visualization Type:-*** Heatmap (Seaborn)

• ***Purpose:-*** To display the percentage share and proportional breakdown of male and female workers across different labor categories using a clear matrix format, avoiding standard bar charts.

• ***Insight:-*** This heatmap clearly highlights gender concentration within specific job roles, showing the exact percentage split between male and female workforce participation for each category.

```python
# Create cross-tabulation of labour category vs gender (percentage normalized)
gender_cat_ct = pd.crosstab(
    df['category_of_labour_(labour_category)'], 
    df['gender_of_labour_(gender)'], 
    normalize='index'
) * 100

# Plot Heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(
    gender_cat_ct, 
    annot=True, 
    fmt='.1f', 
    cmap='Blues', 
    cbar_kws={'label': 'Percentage (%)'}
)
plt.title('Gender Proportion Heatmap Across Labour Categories')
plt.xlabel('Gender')
plt.ylabel('Labour Category')
plt.tight_layout()
plt.show()
```

***info***:- This heatmap provides a matrix-style overview of gender representation across various agricultural roles. By displaying normalized percentages, it enables a direct comparison of how male and female labor allocations differ from one job category to another without relying on standard bar charts.

### (vii). State-wise Agricultural Activity and Engagement Volume

• ***Visualization Type:-Horizontal Lollipop Chart (Matplotlib / Seaborn)

• ***Purpose:-*** To analyze and compare the overall volume of agricultural activities and job engagement records across different states.

• ***Insight:-*** This chart clearly highlights which states record the highest agricultural operations and reporting density, indicating regions with intense farming engagement.

```python
# Calculate total record count per state to measure agricultural activity/engagement volume
state_activity = df['state_name_(state_name)'].value_counts().reset_index()
state_activity.columns = ['State', 'Activity_Count']
state_activity = state_activity.sort_values(by='Activity_Count', ascending=True)

# Plot Horizontal Lollipop Chart
plt.figure(figsize=(11, 6))
plt.hlines(
    y=state_activity['State'], 
    xmin=0, 
    xmax=state_activity['Activity_Count'], 
    color='skyblue', 
    linewidth=2.5
)
plt.plot(
    state_activity['Activity_Count'], 
    state_activity['State'], 
    "o", 
    color='darkblue', 
    markersize=8
)

plt.title('State-wise Agricultural Activity and Engagement Volume')
plt.xlabel('Number of Recorded Activities / Reports')
plt.ylabel('State Name')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

***info***:- This horizontal lollipop chart illustrates the volume of agricultural records and job engagements across various states. By visualizing the frequency of reported data points per region, it identifies which states are primary hubs for agricultural activity and workforce distribution.

### (viii). Wage Variance & Outliers Analysis

• ***Visualization Type:-Box Plot (Seaborn / Matplotlib)

• ***Purpose:-*** To analyze the variance, interquartile range (IQR), and potential outlier values within each labor category rather than relying solely on average wage numbers.

• ***Insight:-*** This plot highlights the dispersion of worker earnings within individual job categories, showing earning consistency and exposing extreme high or low wage values across the dataset.

```python
# Plot Box Plot for wage distribution and outlier detection across labor categories
plt.figure(figsize=(11, 5))
sns.boxplot(
    data=df, 
    x='category_of_labour_(labour_category)', 
    y='average_monthly_wage_(monthly_average_wage)', 
    palette='Set3'
)
plt.title('Wage Spread and Outlier Analysis across Labour Categories')
plt.xlabel('Labour Category')
plt.ylabel('Average Monthly Wage')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

***info***:- This box plot visualizes the statistical spread, median, and outlier values across different agricultural labor categories. By examining the range between the upper and lower quartiles and identifying individual points outside the whiskers, it offers key insights into earning volatility and extreme wage variations within specific job roles.

### (ix). Relationship Between Monthly Wages and Labour Categories

• ***Visualization Type:-Scatter Plot / Strip Plot (Seaborn)

• ***Purpose:-*** To visualize the individual wage distributions and earning clusters across different agricultural labor categories without repeating time-based trend analyses.

• ***Insight:-*** This scatter plot reveals the concentration of wage points within each job role, making it easy to identify earning thresholds and high-value data points across labor categories.

```python
# Plot Scatter (Strip) Plot for Wages across Labour Categories
plt.figure(figsize=(11, 6))
sns.stripplot(
    data=df, 
    x='category_of_labour_(labour_category)', 
    y='average_monthly_wage_(monthly_average_wage)', 
    hue='gender_of_labour_(gender)',
    jitter=True,
    alpha=0.6,
    palette='Set1'
)
plt.title('Distribution and Spread of Monthly Wages across Labour Categories')
plt.xlabel('Labour Category')
plt.ylabel('Average Monthly Wage')
plt.xticks(rotation=30)
plt.legend(title='Gender', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

***info***:- This scatter plot maps individual wage recordings directly against specific labor categories rather than using years. By adding jitter to separate individual observations, it provides a clear view of earning density and wage ranges within each agricultural role without duplicating time-series trend visuals.

### (x). Stacked Bar Chart (Wage Tiers Across Labour Categories)

• ***Visualization Type:-Stacked Bar Chart (Pandas / Matplotlib)

• ***Purpose:-*** To analyze the proportional breakdown of wage categories (Low, Medium, High Wage tiers) within each specific agricultural labor category.

• ***Insight:-*** This chart clearly demonstrates which job roles have the highest concentration of high-paying classifications versus roles dominated by lower wage tiers.

```python
# Calculate percentage proportion of wage tiers per labour category
labour_wage_stack = pd.crosstab(
    df['category_of_labour_(labour_category)'], 
    df['wage_category'], 
    normalize='index'
) * 100

# Plot Stacked Bar Chart
plt.figure(figsize=(11, 6))
labour_wage_stack.plot(
    kind='bar', 
    stacked=True, 
    colormap='viridis', 
    figsize=(11, 6)
)
plt.title('Proportion of Wage Tiers Across Labour Categories')
plt.xlabel('Labour Category')
plt.ylabel('Percentage (%)')
plt.legend(title='Wage Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

***info***- This stacked bar chart provides a normalized percentage view of how different wage categories are distributed across individual labor roles. It highlights occupational economic tiers, showing which specific agricultural activities offer higher financial returns.

### (xi). Violin Plot (Wage Density across Labour Categories)

• ***Visualization Type:-Violin Plot (Seaborn)

• ***Purpose:-*** To visualize both the statistical summary (medians, quartiles) and the full probability density/shape of wages within each labor category.

• ***Insight:-*** Unlike standard box plots, this reveals multi-modal distributions (e.g., if wages cluster around two different peaks within the same job role).

```python
plt.figure(figsize=(11, 6))
sns.violinplot(
    data=df, 
    x='category_of_labour_(labour_category)', 
    y='average_monthly_wage_(monthly_average_wage)', 
    palette='Pastel1',
    inner='quartile'
)
plt.title('Violin Plot of Wage Density and Spread Across Labour Categories')
plt.xlabel('Labour Category')
plt.ylabel('Average Monthly Wage')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

***info***- This violin plot combines box plot statistics with a smoothed density trace. By showing the full distribution shape of earnings across different agricultural job roles, it uncovers hidden variations and concentration patterns in worker wages.

<h2 style="font-size:24px">Objective of the Analysis</h2>

<p style="font-size:16px;text-align:justify">The objective of this project was to analyze agricultural wage data to identify key regional, demographic, and temporal factors associated with wage variations and labor trends.Through systematic data preprocessing, exploratory data analysis (EDA), and comprehensive visual analytics, the study aims to uncover meaningful patterns that can support a better understanding of rural labor economics, wage disparities, and agricultural income distribution.</p>

<h2 style="font-size:24px">Data Preparation Summary</h2>

<p style="font-size:16px;text-align:justify">The original agricultural wage dataset consisted of 500 records with mixed numerical and categorical features. Several preprocessing steps were undertaken to enhance data quality and analytical reliability, including handling missing values, replacing invalid coordinate entries (0.0 values) via forward fill, and filling zero monthly average wages using column medians.<br>
Feature engineering steps such as wage categorization (Low, Medium, High wage groups), correcting annual average wage calculations, and multi-variable group-by aggregations improved interpretability and enabled more structured regional comparisons. After preprocessing, the final dataset formed a robust foundation for rural labor economics and wage trend analysis.</p>

<h2 style="font-size:24px">Exploratory Data Analysis Overview</h2>

<p style="font-size:16px;text-align:justify">Initial exploration revealed significant regional and state-wise variations in agricultural wages, highlighting the importance of evaluating average wage rates and income disparities rather than absolute figures alone.<br>EDA highlighted distinct trends in wage progression over the years, differences in compensation between men and women across various tasks, and notable wage variations across different labor categories and skill types, indicating key economic and demographic contributors to rural labor distribution.</p>

<h2 style ="font-size:24px">Key Analytical Insights from Visual Exploration</h2>

<h2 style ="font-size:20px">(i) Regional Disparities</h2>

<p style="font-size:16px;text-align:justify">State-wise wage analysis highlights significant geographical variations, where states like Kerala and Uttarakhand record higher average monthly wages compared to regions like Himachal Pradesh and Gujarat.</p>


<h2 style ="font-size:20px">(ii) Temporal Wage Growth</h2>

<p style="font-size:16px;text-align:justify">The year-over-year trend demonstrates a steady and progressive upward movement in agricultural wages from 2010 to 2020, reflecting inflationary adjustments and shifting rural labor economics.</p>

<h2 style ="font-size:20px">(iii) Gender Wage Gaps </h2>

<p style="font-size:16px;text-align:justify">Across various labor classifications and operating centers, male workers consistently exhibit slightly higher average monthly wage rates than female workers.</p>

<h2 style ="font-size:20px">(iv) Labor Classification Impact </h2>

<p style="font-size:16px;text-align:justify">Specialized labor categories and skilled agricultural operations command considerably higher compensation tiers compared to routine general farming tasks.</p>

<h2 style ="font-size:24px">Overall Findings and Conclusion</h2>

<p style="font-size:16px;text-align:justify">Overall, this analysis demonstrates that agricultural wage rates are primarily driven by geographical location, labor classification, and gender dynamics. While regional economic conditions and seasonal variations contribute to differences in pay rates, wage disparities are notably influenced by labor types, skill levels, and state-wise policies.<br>
The findings emphasize that agricultural earnings are not determined by a single factor but rather by the interaction of multiple socio-economic dimensions, with regional economic factors acting as core drivers and specialized labor classifications serving as key determinants of higher compensation.</p>

<h2 style ="font-size:24px">💡Recommendation</h2>

<p style="font-size:16px;text-align:justify">Focus on improving daily wage rates for low-income agricultural workers, especially female laborers and workers in low-wage states. By using data-driven wage tracking, governments and policymakers can ensure fair pay, reduce income gaps, and support rural livelihoods more effectively.</p>
